```python
def get_knn(train_features, query_features, k, metric):
    
    # Criar o modelo
    nbrs = NearestNeighbors(n_neighbors=k, metric=metric)
    
    # Ajustar o modelo nos dados de treinamento
    nbrs.fit(train_features)
    
    # Encontrar os k vizinhos mais próximos
    nbrs_distances, nbrs_indices = nbrs.kneighbors(
        query_features,
        n_neighbors=k,
        return_distance=True
    )
    
    # Criar tabela com os vizinhos encontrados
    df_res = train_features.iloc[nbrs_indices[0]].copy()
    df_res['distance'] = nbrs_distances[0]
    
    return df_res
```

#### Regressão (com Regressão Linear)

```python
class MyLinearRegression:
    
    def __init__(self):
        
        self.weights = None
    
    def fit(self, X, y):
        
        # somando as unidades
        X2 = np.append(np.ones([len(X), 1]), X, axis=1)
        self.weights = np.linalg.inv(X2.T.dot(X2)).dot(X2.T).dot(y)

    def predict(self, X):
        
        # somando as unidades
        X2 = np.append(np.ones([len(X), 1]), X, axis=1)
        y_pred = X2.dot(self.weights)
        
        return y_pred
```

```python
def eval_regressor(y_true, y_pred):
    
    rmse = math.sqrt(sklearn.metrics.mean_squared_error(y_true, y_pred))
    print(f'REQM: {rmse:.2f}')
    
    r2_score = math.sqrt(sklearn.metrics.r2_score(y_true, y_pred))
    print(f'R2: {r2_score:.2f}')    
```

```python
X = df[['age', 'gender', 'income', 'family_members']].to_numpy()
y = df['insurance_benefits'].to_numpy()

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=12345)

lr = MyLinearRegression()

lr.fit(X_train, y_train)
print(lr.weights)

y_test_pred = lr.predict(X_test)
eval_regressor(y_test, y_test_pred)
```

```python
scaler = MaxAbsScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

lr_scaled = MyLinearRegression()

lr_scaled.fit(X_train_scaled, y_train)

y_test_pred_scaled = lr_scaled.predict(X_test_scaled)

eval_regressor(y_test, y_test_pred_scaled)
```

Ao comparar a regressão linear com as features originais e com as features escaladas pelo MaxAbsScaler, o REQM permaneceu o mesmo, 0,34, e o R² também permaneceu 0,66 nos dois casos. Isso confirma empiricamente que a solução analítica exata da regressão linear é invariante ao reescalonamento das features. O MaxAbsScaler realiza apenas um reescalonamento linear invertível de cada feature. Dessa forma, as predições e as métricas do modelo permanecem inalteradas após o escalonamento.

## Teste de regressão linear com ofuscação de dados

```python
rng = np.random.default_rng(seed=42)
P = rng.random(size=(X.shape[1], X.shape[1]))


# 2. Verificar se P é invertível
while np.isclose(np.linalg.det(P), 0):
    P = rng.random(size=(X.shape[1], X.shape[1]))


# 3. Ofuscar os dados usando a matriz P
X_obfuscated = X.dot(P)


# 4. Usar X_obfuscated como a nova matriz de características
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=12345
)

X_obf_train, X_obf_test, y_obf_train, y_obf_test = train_test_split(
    X_obfuscated,
    y,
    test_size=0.3,
    random_state=12345
)


# Modelo com dados originais
model_original = LinearRegression()
model_original.fit(X_train, y_train)

y_pred_original = model_original.predict(X_test)


# Modelo com dados ofuscados
model_obfuscated = LinearRegression()
model_obfuscated.fit(X_obf_train, y_obf_train)

y_pred_obfuscated = model_obfuscated.predict(X_obf_test)


# Comparar REQM
rmse_original = np.sqrt(mean_squared_error(y_test, y_pred_original))
rmse_obfuscated = np.sqrt(mean_squared_error(y_obf_test, y_pred_obfuscated))

print(f"REQM original:    {rmse_original:.10f}")
print(f"REQM ofuscado:    {rmse_obfuscated:.10f}")
```

A transformação de ofuscação não alterou o desempenho da Regressão Linear. Os valores de REQM dos dados originais e ofuscados foram exatamente iguais (0,3435565089), mostrando que, mesmo com os dados das características transformados, o modelo conseguiu manter as mesmas previsões. Isso confirma que a ofuscação escolhida protege os dados sem prejudicar a qualidade da regressão.
